## Import Libraries

```python
import GEOparse
import os
```

## Create Download Directory

```python
os.makedirs("data/raw/GSE178341", exist_ok=True)
```

## Fetch GEO Series Metadata

```python
gse = GEOparse.get_GEO(geo="GSE178341", destdir="data/raw/GSE178341")
```

## Download Supplementary Files

```python
gse.download_supplementary_files(directory="data/raw/GSE178341")
```

## Download GSE-level Supplementary Files

```python
import urllib.request

files = [
    "GSE178341_crc10x_full_c295v4_submit.h5",
    "GSE178341_crc10x_full_c295v4_submit_cluster.csv.gz",
    "GSE178341_crc10x_full_c295v4_submit_metatables.csv.gz"
]

base_url = "https://ftp.ncbi.nlm.nih.gov/geo/series/GSE178nnn/GSE178341/suppl/"

for f in files:
    dest = f"data/raw/GSE178341/{f}"
    print(f"Downloading {f}...")
    urllib.request.urlretrieve(base_url + f, dest)
    print(f"Done: {dest}")
```

## Verify Downloads

```python
import os

files = [
    "GSE178341_crc10x_full_c295v4_submit.h5",
    "GSE178341_crc10x_full_c295v4_submit_cluster.csv.gz",
    "GSE178341_crc10x_full_c295v4_submit_metatables.csv.gz"
]

for f in files:
    path = f"data/raw/GSE178341/{f}"
    size_mb = os.path.getsize(path) / 1e6
    print(f"{f}: {size_mb:.1f} MB")
```
